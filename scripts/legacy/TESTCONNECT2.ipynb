```python
import os, sys
import pyodbc
import pandas as pd

server   = os.environ.get("K2_SERVER", "43.254.133.123")
database = os.environ.get("K2_DB", "HPCOM7")
username = os.environ.get("K2_USER")
password = os.environ.get("K2_PWD")

if not username or not password:
    raise SystemExit('ต้องตั้ง env K2_USER / K2_PWD ก่อน')

conn = pyodbc.connect(
    f"DRIVER={{ODBC Driver 18 for SQL Server}};"
    f"SERVER={server};DATABASE={database};"
    f"UID={username};PWD={password};TrustServerCertificate=yes;"
)
print("Connected!")

df = pd.read_sql("SELECT TOP (1000) * FROM rpt.dim_item_itec", conn)
print(df)
```
